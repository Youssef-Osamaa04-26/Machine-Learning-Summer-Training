```python
# File: 01_data_cleaning.ipynb
# Team: Mohamed Qorany + Yahia
# Task: Data Cleaning
# ----------------------------------------

import pandas as pd
import numpy as np

print("=" * 60)
print("DATA CLEANING")
print("=" * 60)

# -----------------------------------------
# 1. Load Raw Data

df = pd.read_csv('../1_Data/raw_data.csv', sep='\t')

print(f"\n Data Loaded")
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print("\nFirst 5 rows:")
print(df.head())

# -----------------------------------------
# 2. Check Missing Values

print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)
missing = df.isnull().sum()
print(missing)
print(f"\nTotal Missing: {missing.sum()}")

# -----------------------------------------
# 3. Check Duplicates

print("\n" + "=" * 60)
print("DUPLICATES")
print("=" * 60)
duplicates = df.duplicated().sum()
print(f"Duplicate Rows: {duplicates}")

# -----------------------------------------
# 4. Check Data Types

print("\n" + "=" * 60)
print("DATA TYPES")
print("=" * 60)
print(df.dtypes)

# -----------------------------------------
# 5. Basic Statistics

print("\n" + "=" * 60)
print("BASIC STATISTICS")
print("=" * 60)
print(df.describe())

# -----------------------------------------
# 6. Save Cleaned Data

df.to_csv('../1_Data/cleaned_data.csv', index=False, encoding='utf-8')
print("\n" + "=" * 60)
print("cleaned_data.csv SAVED")
print("=" * 60)

# -----------------------------------------
# 7. Verify

df_check = pd.read_csv('../1_Data/cleaned_data.csv')
print(f"\nVerification:")
print(f"Shape: {df_check.shape}")
print(f"Columns: {df_check.columns.tolist()}")
print(df_check.head())
```
