```python
# File: 02_eda_analysis.ipynb
# Team: Mohamed Qorany + Yahia
# Task: Exploratory Data Analysis (EDA)
# ------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 60)
print("EDA ANALYSIS")
print("=" * 60)

# -----------------------------------------
# 1. Load Cleaned Data

df = pd.read_csv('../1_Data/cleaned_data.csv')

print(f"\n Data Loaded")
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print(df.head())

# -----------------------------------------
# 2. Summary Statistics

print("\n" + "=" * 60)
print("SUMMARY STATISTICS")
print("=" * 60)
print(df.describe())

# -----------------------------------------
# 3. Chart 1: Average Price by Location

print("\n Creating Chart 1: Price by Location...")

plt.figure(figsize=(10, 6))
avg_price = df.groupby('Location')['Price_EGP'].mean().sort_values(ascending=False)
bars = plt.bar(avg_price.index, avg_price.values, color='steelblue')
plt.title('Average Property Price by Location', fontsize=14, fontweight='bold')
plt.xlabel('Location', fontsize=12)
plt.ylabel('Average Price (EGP)', fontsize=12)
plt.xticks(rotation=45)

for bar, val in zip(bars, avg_price.values):
    plt.text(bar.get_x() + bar.get_width()/2, val, 
             f'{val:,.0f}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig('chart_1_price_by_location.png', dpi=150, bbox_inches='tight')
plt.show()
print("chart_1 saved")

# -----------------------------------------
# 4. Chart 2: Property Type Distribution

print("\n Creating Chart 2: Property Type Pie...")

plt.figure(figsize=(8, 8))
type_counts = df['Property_Type'].value_counts()
colors = ['#3498db', '#e74c3c', '#2ecc71', '#f39c12']
plt.pie(type_counts.values, labels=type_counts.index, 
        autopct='%1.1f%%', colors=colors, startangle=90)
plt.title('Property Type Distribution', fontsize=14, fontweight='bold')
plt.savefig('chart_2_property_type_pie.png', dpi=150, bbox_inches='tight')
plt.show()
print("chart_2 saved")

# -----------------------------------------
# 5. Chart 3: Price Distribution

print("\n Creating Chart 3: Price Histogram...")

plt.figure(figsize=(10, 6))
plt.hist(df['Price_EGP'], bins=30, color='green', alpha=0.7, edgecolor='black')
plt.axvline(df['Price_EGP'].mean(), color='red', linestyle='--', 
            linewidth=2, label=f'Mean: {df["Price_EGP"].mean():,.0f}')
plt.title('Price Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Price (EGP)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.legend()
plt.tight_layout()
plt.savefig('chart_3_price_histogram.png', dpi=150, bbox_inches='tight')
plt.show()
print("chart_3 saved")

# -----------------------------------------
# 6. Chart 4: Area vs Price

print("\n Creating Chart 4: Area vs Price...")

plt.figure(figsize=(10, 6))
plt.scatter(df['Area_SqM'], df['Price_EGP'], alpha=0.5, color='purple', s=50)
plt.title('Area vs Price', fontsize=14, fontweight='bold')
plt.xlabel('Area (SqM)', fontsize=12)
plt.ylabel('Price (EGP)', fontsize=12)
plt.tight_layout()
plt.savefig('chart_4_area_vs_price.png', dpi=150, bbox_inches='tight')
plt.show()
print("chart_4 saved")

# ------------------------------------------
# 7. Chart 5: Correlation Heatmap

print("\n Creating Chart 5: Correlation Heatmap...")

plt.figure(figsize=(10, 8))
numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
            fmt='.3f', linewidths=0.5, square=True)
plt.title('Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('chart_5_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print("chart_5 saved")

# -----------------------------------------
# 8. Final Summary

print("\n" + "=" * 60)
print("EDA COMPLETE")
print("=" * 60)
print("\n Generated Files:")
print("  - chart_1_price_by_location.png")
print("  - chart_2_property_type_pie.png")
print("  - chart_3_price_histogram.png")
print("  - chart_4_area_vs_price.png")
print("  - chart_5_correlation_heatmap.png")
print("\n All charts saved!")
```
